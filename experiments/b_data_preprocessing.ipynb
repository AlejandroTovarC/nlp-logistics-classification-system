## Objective

The purpose of this notebook is to perform pre-processing of the customer support ticket dataset to get it ready for feature engineering and model training. Pre-processing steps include tokenizing, converting to lowercase, lemmatizing, removing stop words, and stripping sentance punctuation.

The output of this notebook is a csv file with the raw 'Combined Text' replaced with pre-processed/cleaned text.

## Pre-processing

Function for pre-processing data:
- Tokenizes the text 
- Converts text to lowercase
  - The dataset appears to already contain text that is all lowercase, this step is being added to the function to account for the potential of future capitalized text.
- lemmatizes the text to obtain root word
- removes stop words
- strips punctuation from beginning and end of each token

```python
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()
tokenizer = RegexpTokenizer(r'[A-Za-z]+')
```

```python
def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    return wordnet.NOUN


def preprocess_text(text):

    tokens = tokenizer.tokenize(text.lower())

    tagged = pos_tag(tokens)

    tokens = [lemmatizer.lemmatize(word, get_wordnet_pos(tag))
            for word, tag in tagged if word not in stop_words]

    return " ".join(tokens)

train_df['processed_text'] = train_df['Combined Text'].apply(preprocess_text)

train_df.head()
```
