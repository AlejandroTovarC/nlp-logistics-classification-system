## Imports

```python
import pandas as pd
import sklearn
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction.text import CountVectorizer
```

Retrieve pre-processed training data

```python
train_df = pd.read_csv("../data/nlp_dataset_train_pre-processed.csv")
```

## Set TF-IDF function using sklearn

Function TF_IDF(df, text):
- Inputs
  - df = df containing the text to process;
  - text = reference to text containing column.
- Initalizes the vectorizer
- applies the vectorizer to the input text
- creates a dataframe from the resulting output, with extracted feature names are used for the column headers
- merges the new TF-IDF data frame back into the original dataframe. 

```python
def TF_IDF(df, text):
    df = df
    text = text
    
    tfidf_vectorizer = TfidfVectorizer(stop_words = 'english', smooth_idf = True, norm = 'l2')
    tfidf_vector = tfidf_vectorizer.fit_transform(text)
    feature_names = tfidf_vectorizer.get_feature_names_out()
    
    tfidf_df = pd.DataFrame(tfidf_vector.toarray(), columns = feature_names)
    tfidf_df = df.join(tfidf_df, how = 'outer')
    return tfidf_df
```

Apply the function to the training dataset

```python
#Function Testing on Training Set
df = train_df
text = train_df['Combined Text']
tfidf_df = TF_IDF(df, text)
tfidf_df.head(6)
```

## Set BoW Function using sklearn

Function BoW(df, text):
- Inputs
  - df = df containing the text to process;
  - text = reference to text containing column.
- Initalizes the vectorizer
- applies the vectorizer to the input text
- creates a dataframe from the resulting output
- merges the new BoW data frame back into the original dataframe

```python
def BoW(df, text):
    df = df
    text = text

    bow_vectorizer = CountVectorizer()
    bow_vector = bow_vectorizer.fit_transform(text)
    feature_names = bow_vectorizer.get_feature_names_out()

    bow_df = pd.DataFrame(bow_vector.toarray(), columns=feature_names)
    bow_df = df.join(bow_df, how = 'outer')
    return bow_df
```

Apply the function to the training dataset

```python
#Function Testing on Training Set
df = train_df
text = train_df['Combined Text']
bow_df = BoW(df, text)
bow_df.head(6)
```
