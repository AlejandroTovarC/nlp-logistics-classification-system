## Imports

```python
import pandas as pd
import sklearn
from sklearn.feature_extraction.text import TfidfVectorizer
```

Retrieve pre-processed training data

```python
train_df = pd.read_csv("../data/nlp_dataset_train_pre-processed.csv")
```

## Set TF-IDF function using sklearn

Function TF_IDF(text):
- Initalizes the vectorizer
- applies the vectorizer to the input text
- creates a dataframe from the resulting output, with extracted feature names are used for the column headers

```python
def TF_IDF(text):
    text = text
    tfidf_vectorizer = TfidfVectorizer(stop_words = 'english', smooth_idf = True, norm = 'l2')
    tfidf_vector = tfidf_vectorizer.fit_transform(text)
    feature_names = tfidf_vectorizer.get_feature_names_out()
    tf_idf_df = pd.DataFrame(tfidf_vector.toarray(), columns = feature_names)
    return tf_idf_df
```

Apply the function to the training dataset

```python
text = train_df['Combined Text']
tf_idf_df = TF_IDF(text)
tf_idf_df.head(6)
```
