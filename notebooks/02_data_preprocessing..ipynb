## Objective

The purpose of this notebook is to perform pre-processing of the customer support ticket dataset to get it ready for feature engineering and model training. Pre-processing steps include tokenizing, converting to lowercase, lemmatizing, removing stop words, and stripping sentance punctuation.

The output of this notebook is a csv file with the raw 'Combined Text' replaced with pre-processed/cleaned text.

## Imports

```python
import pandas as pd
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize, RegexpTokenizer, sent_tokenize
from nltk.corpus import stopwords
import string
```

## Pre-processing

Function for pre-processing data:
- Tokenizes the text
- Converts text to lowercase
  - The dataset appears to already contain text that is all lowercase, this ttep is being added to the function to account for the potential of future capitalized text.
- lemmatizes the text to obtain root word
- removes stop words
- strips punctuation from beginning and end of each token

```python
def preprocess_text(text):
    stop_words = set(stopwords.words('english'))
    lemmatizer = WordNetLemmatizer()
    tokenizer = RegexpTokenizer(r'\s+', gaps=True)
    
    tokens = tokenizer.tokenize(text)  # Converts the text into tokens
    tokens = [token.lower() for token in tokens]  # Reverts all letters to lower case
    tokens = [lemmatizer.lemmatize(token) for token in tokens] #Lemmatizes the word to its root
    tokens = [token for token in tokens if token not in stop_words] # Removes stop words
    tokens = [token.strip(string.punctuation) for token in tokens] # Strips punctuation from beginning and end of token
    
    return ' '.join(tokens) # rejoins tokens in to a single string
```

Run Function on dataframe column - Replaces column with pre-processed text

```python
df['Combined Text'] = df['Combined Text'].apply(preprocess_text)
df.head(10)
```
