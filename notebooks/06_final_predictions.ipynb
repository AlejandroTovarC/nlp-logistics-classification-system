## Objective

This notebook demonstrates how the trained machine learning models can be used to classify new customer support tickets into their predicted categories.

The prediction pipeline includes:

- Loading the trained model(s) and saved vectorizer(s)
- Preparing new customer support messages
- Transforming the text using the saved vectorizer
- Predicting the ticket category
- Displaying the predicted ticket labels
- Comparing predictions from the trained models
- Testing the models with multiple example customer support tickets

## Import Libraries

```python
import pandas as pd
import joblib
```

## Load Saved Model and Vectorizer

```python
model = joblib.load("../models/tfidf_naive_bayes_model.pkl")
vectorizer = joblib.load("../models/tfidf_vectorizer.pkl")
```

## Create New Support Tickets

```python
new_tickets = [
    "My package has not arrived yet.",
    "I received a damaged item.",
    "I would like a refund.",
    "I forgot my account password.",
    "The payment was charged twice."
]
```

## Transform the Text

```python
X_new = vectorizer.transform(new_tickets)
```

## Make Predictions

```python
predictions = model.predict(X_new)
```

## Display Results

```python
results = pd.DataFrame({
    "Customer Message": new_tickets,
    "Predicted Ticket Type": predictions
})

results
```

## Conclusion

The trained NLP model successfully classified new customer support messages into their predicted ticket categories. This demonstrates how the saved machine learning model can be deployed to automatically categorize incoming support tickets, reducing manual effort and improving customer service efficiency.
